```python
stats_num = pd.DataFrame(index=current_numerical_variables)

stats_num["Nombre de valeurs"] = data_no_duplicates[current_numerical_variables].count()
stats_num["Moyenne"] = data_no_duplicates[current_numerical_variables].mean()
stats_num["Médiane"] = data_no_duplicates[current_numerical_variables].median()
stats_num["Écart-type"] = data_no_duplicates[current_numerical_variables].std()
stats_num["Minimum"] = data_no_duplicates[current_numerical_variables].min()
stats_num["Q1"] = data_no_duplicates[current_numerical_variables].quantile(0.25)
stats_num["Q3"] = data_no_duplicates[current_numerical_variables].quantile(0.75)
stats_num["Maximum"] = data_no_duplicates[current_numerical_variables].max()

stats_num["Étendue"] = stats_num["Maximum"] - stats_num["Minimum"]
stats_num["Écart interquartile"] = stats_num["Q3"] - stats_num["Q1"]

stats_num.round(2).T
```

```python
import numpy as np

stats_num["Différence moyenne-médiane"] = abs(
    stats_num["Moyenne"] - stats_num["Médiane"]
)


stats_num["Différence relative (%)"] = (
    stats_num["Différence moyenne-médiane"] / abs(stats_num["Médiane"])
) * 100


stats_num["Borne inférieure"] = (
    stats_num["Q1"] - 1.5 * stats_num["Écart interquartile"]
)

stats_num["Borne supérieure"] = (
    stats_num["Q3"] + 1.5 * stats_num["Écart interquartile"]
)

nb_aberrantes = {}

for var in current_numerical_variables:
    borne_inf = stats_num.loc[var, "Borne inférieure"]
    borne_sup = stats_num.loc[var, "Borne supérieure"]

    nb_aberrantes[var] = (
        ((data_no_duplicates[var] < borne_inf) |
         (data_no_duplicates[var] > borne_sup))
    ).sum()

stats_num["Nombre de valeurs aberrantes"] = pd.Series(nb_aberrantes)

stats_num["Remplacement conseillé"] = np.where(
    (stats_num["Différence relative (%)"] <= 10)
    & (stats_num["Nombre de valeurs aberrantes"] == 0),
    "Moyenne",
    "Médiane"
)

display(stats_num.round(2))
```

```python
df_clean = data_no_duplicates.copy()

print("Valeurs manquantes avant imputation :")
display(df_clean.isnull().sum())

colonnes_moyenne = stats_num[
    stats_num["Remplacement conseillé"] == "Moyenne"
].index.tolist()

colonnes_mediane = stats_num[
    stats_num["Remplacement conseillé"] == "Médiane"
].index.tolist()

colonnes_moyenne = [
    col for col in colonnes_moyenne
    if col in df_clean.columns
]

colonnes_mediane = [
    col for col in colonnes_mediane
    if col in df_clean.columns
]

df_clean[colonnes_moyenne] = df_clean[colonnes_moyenne].fillna(
    df_clean[colonnes_moyenne].mean()
)

df_clean[colonnes_mediane] = df_clean[colonnes_mediane].fillna(
    df_clean[colonnes_mediane].median()
)

for col in ["A", "B"]:
    if col in df_clean.columns and df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )

print("\nValeurs manquantes après imputation :")
display(df_clean.isnull().sum())
```

```python
def interpreter_p_value(p_value, alpha=0.05):
    if p_value < alpha:
        return "On rejette H0 : le résultat est statistiquement significatif."
    else:
        return "On ne rejette pas H0 : le résultat n’est pas statistiquement significatif."
```

```python
alpha = 0.05
```

```python
#H0 : Il n'existe pas de corrélation significative entre X3 et X7.
#H1 : Il existe une corrélation significative entre X3 et X7.
data_corr = df[["X3", "X7"]].dropna()

data_corr
```

```python
corr, p_value_corr = stats.pearsonr(data_corr["X3"], data_corr["X7"])

print("Coefficient de corrélation :", corr)
print("p-value :", p_value_corr)
print(interpreter_p_value(p_value_corr))
```

```python
#H0 : Il n'existe pas de corrélation significative entre X7 et X10.
#H1 : Il existe une corrélation significative entre X7 et X10.
data_corr_yield_protein = df[["X7", "X10"]].dropna()

corr_yield_protein, p_value_yield_protein = stats.pearsonr(
    data_corr_yield_protein["X7"],
    data_corr_yield_protein["X10"]
)

print("Coefficient de corrélation :", corr_yield_protein)
print("p-value :", p_value_yield_protein)
print(interpreter_p_value(p_value_yield_protein))
```

```python
variables_quantitatives = [
    "X1","X2","X3","X4","X5",
    "X6","X7","X8","X9","X10","X11"
]

X = df[variables_quantitatives]
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2)
components = pca.fit_transform(X_scaled)

df_pca = pd.DataFrame(components, columns=["PC1", "PC2"])

df_pca["A"] = df["A"]
df_pca["B"] = df["B"]

print("Variance expliquée :", pca.explained_variance_ratio_)

plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df_pca,
    x="PC1",
    y="PC2",
    hue="A",
    palette=["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728"]
)

plt.title("Projection des observations par ACP")
plt.xlabel("Composante principale 1 (PC1)")
plt.ylabel("Composante principale 2 (PC2)")
plt.legend(title="Traitement A", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()

contributions_vars = pd.DataFrame(
    pca.components_,
    columns=variables_quantitatives,
    index=["PC1", "PC2"]
)

print("\nContribution des variables :")
display(contributions_vars)
```

```python
pca_3 = PCA(n_components=3)

X_pca_3 = pca_3.fit_transform(X_scaled)

print("Variance expliquée :", pca_3.explained_variance_ratio_)
print(
    "Variance expliquée cumulée :",
    pca_3.explained_variance_ratio_.sum()
)

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    X_pca_3[:, 0],
    X_pca_3[:, 1],
    X_pca_3[:, 2]
)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")
ax.set_title("Projection 3D des observations par ACP")

plt.tight_layout()
plt.show()

contributions_vars_3 = pd.DataFrame(
    pca_3.components_,
    columns=variables_quantitatives,
    index=["PC1", "PC2", "PC3"]
)

print("\nContribution des variables :")
display(contributions_vars_3)
```
